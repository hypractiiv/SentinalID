## Step 1: Upload document photo and selfie

```python
from google.colab import files
print("Upload the document photo:")
doc_upload = files.upload()
doc_path = list(doc_upload.keys())[0]

print("Upload the selfie/live photo:")
selfie_upload = files.upload()
selfie_path = list(selfie_upload.keys())[0]
```

## Step 2: Face verification function

```python
from deepface import DeepFace

def verify_faces(doc_image_path, selfie_image_path):
    result = DeepFace.verify(
        img1_path=doc_image_path,
        img2_path=selfie_image_path,
        model_name="ArcFace"
    )
    return {
        "match": result["verified"],
        "distance": round(result["distance"], 4),
        "similarity_score": round((1 - result["distance"]) * 100, 2)
    }

face_result = verify_faces(doc_path, selfie_path)
face_result
```

## Step 3: Visual sanity check
